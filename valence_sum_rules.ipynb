```python
import lhapdf
from scipy.integrate import fixed_quad
import numpy as np

p = lhapdf.mkPDF("NNPDF40_nnlo_as_01180", 0)

Q2 = 10.0  # fixed Q^2, we can check the sum rules at any Q^2, they should hold for all Q^2

def _uv_scalar(x, Q2):
    u = p.xfxQ2(2, x, Q2) / x
    ubar = p.xfxQ2(-2, x, Q2) / x
    return u - ubar

def _dv_scalar(x, Q2):
    d = p.xfxQ2(1, x, Q2) / x
    dbar = p.xfxQ2(-1, x, Q2) / x
    return d - dbar

# vectorized versions
uv = np.vectorize(_uv_scalar, excluded={"Q2"})
dv = np.vectorize(_dv_scalar, excluded={"Q2"})

x_min = 1e-10
x_max = 1.0

# integrate u_v(x, Q^2) from 0 to 1, same for d_v(x, Q^2)
uv_sr, _ = fixed_quad(uv, x_min, x_max, args=(Q2,), n=1000)
dv_sr, _ = fixed_quad(dv, x_min, x_max, args=(Q2,), n=1000)
print("Valence sum rule (uv)", uv_sr)
print("Valence sum rule (dv)", dv_sr)
```
