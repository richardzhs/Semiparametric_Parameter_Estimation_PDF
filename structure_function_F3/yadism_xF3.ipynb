```python
# theory_card = {
#     "PTO": 2,
#     "FNS": "ZM-VFNS",
#     "DAMP": 0,
#     "IC": 0,
#     "IB": 0,
#     "ModEv": "TRN",
#     "ModSV": "unvaried",
#     "XIR": 1.0,
#     "XIF": 1.0,
#     "NfFF": 5,
#     "MaxNfAs": 5,
#     "MaxNfPdf": 5,
#     "Q0": 1.65,
#     "alphas": 0.118,
#     "Qref": 91.2,
#     "nf0": 4,
#     "nfref": 5,
#     "QED": 0,
#     "alphaqed": 0.007496252,
#     "Qedref": 1.777,
#     "SxRes": 0,
#     "SxOrd": "LL",
#     "HQ": "POLE",
#     "mc": 1.51,
#     "Qmc": 1.51,
#     "kcThr": 1.0,
#     "mb": 4.92,
#     "Qmb": 4.92,
#     "kbThr": 1.0,
#     "mt": 172.5,
#     "Qmt": 172.5,
#     "ktThr": 1.0,
#     "CKM": "0.97428 0.22530 0.003470 0.22520 0.97345 0.041000 0.00862 0.04030 0.999152",
#     "MZ": 91.1876,
#     "MW": 80.398,
#     "GF": 1.1663787e-05,
#     "SIN2TW": 0.23126,
#     "TMC": 0,
#     "MP": 0.938,
#     "global_nx": 0,
#     "EScaleVar": 1,
#     "kDIScThr": 1.0,
#     "kDISbThr": 1.0,
#     "kDIStThr": 1.0,
#     "n3lo_cf_variation": 0,
# }

theory_card = {
    "PTO": 2,
    "FNS": "FFNS",
    "DAMP": 0,
    "IC": 1,
    "IB": 0,
    "ModEv": "TRN",
    "ModSV": "unvaried",
    "XIR": 1.0,
    "XIF": 1.0,
    "NfFF": 5,
    "MaxNfAs": 5,
    "MaxNfPdf": 5,
    "Q0": 1.65,
    "alphas": 0.118,
    "Qref": 91.2,
    "nf0": 4,
    "nfref": 5,
    "QED": 0,
    "alphaqed": 0.007496252,
    "Qedref": 1.777,
    "SxRes": 0,
    "SxOrd": "LL",
    "HQ": "POLE",
    "mc": 1.51,
    "Qmc": 1.51,
    "kcThr": 1.0,
    "mb": 4.92,
    "Qmb": 4.92,
    "kbThr": 4.0,
    "mt": 172.5,
    "Qmt": 172.5,
    "ktThr": 1.0,
    "CKM": "0.97428 0.22530 0.003470 0.22520 0.97345 0.041000 0.00862 0.04030 0.999152",
    "MZ": 91.1876,
    "MW": 80.398,
    "GF": 1.1663787e-05,
    "SIN2TW": 0.23126,
    "TMC": 1,
    "MP": 0.938,
    "global_nx": 0,
    "EScaleVar": 1,
    "kDIScThr": 1.0,
    "kDISbThr": 4.0,
    "kDIStThr": 1.0,
    "n3lo_cf_variation": 0,
}

observables_card = {
    "PolarizationDIS": 0.0,
    "ProjectileDIS": "electron",
    "PropagatorCorrection": 0.0,
    "TargetDIS": "proton",
    "interpolation_is_log": True,
    "interpolation_polynomial_degree": 4,
    "interpolation_xgrid": interpolation.lambertgrid(60).tolist(),
    "observables": {"F3_total": []},
    "prDIS": "NC",
    "NCPositivityCharge": None,
}
yadism.log.silent_mode = True


def compute():
    """Compute yadism prediction for all experimental bins."""
    obs_card = copy.deepcopy(observables_card)
    # collect all kinematics
    curobs = data.apply(
        lambda dat: {"x": dat["x"], "Q2": dat["Q2"]},
        axis=1,
    )
    obs_card["observables"]["F3_total"] = list(curobs.to_dict().values())
    out = yadism.run_yadism(theory_card, obs_card)
    # let's dump the grid, which we can reuse
    dump_pineappl_to_file(
        out, "HERA_F3.pineappl.lz4", "F3_total"
    )
```

```python
compute()
```
